```python
import pandas as pd
import re
import os

# Define a function to do all the cleaning steps at once
def clean_dataframe(df):
    # Step 2: Replace missing values with empty strings
    df = df.fillna('')

    # Apply cleaning to the three text columns
    cols_to_clean = ['business_name', 'business_address', 'country']

    for col in cols_to_clean:
        if col in df.columns:
            # Step 3: Convert to string and lowercase
            df[col] = df[col].astype(str).str.lower()

            # Step 4: Replace all punctuation (anything that is not a letter, number, or space) with a space
            df[col] = df[col].str.replace(r'[^\w\s]', ' ', regex=True)

            # Step 5: Replace multiple spaces with a single space, and strip edges
            df[col] = df[col].str.replace(r'\s+', ' ', regex=True).str.strip()

    return df

# Step 1: Load the training data (Make sure your Google Drive is mounted first!)
print("Loading data...")
train_s1 = pd.read_csv('dataset/train/train_source1.tsv', sep='\t')
train_s2 = pd.read_csv('dataset/train/train_source2.tsv', sep='\t')
train_s3 = pd.read_csv('dataset/train/train_source3.tsv', sep='\t')

print("Cleaning data...")
# Run the cleaning function on all datasets
clean_s1 = clean_dataframe(train_s1)
clean_s2 = clean_dataframe(train_s2)
clean_s3 = clean_dataframe(train_s3)

# Display a sample to verify it worked
print("\nSample of cleaned Source 1:")
display(clean_s1.head(3))

# Step 6: Save the cleaned versions back to Drive for the next team member
print("Saving cleaned data for Member 2...")
# We save them as TSV to maintain consistency
clean_s1.to_csv('dataset/train/clean_train_source1.tsv', sep='\t', index=False)
clean_s2.to_csv('dataset/train/clean_train_source2.tsv', sep='\t', index=False)
clean_s3.to_csv('dataset/train/clean_train_source3.tsv', sep='\t', index=False)
print("Done!")
```
